```python
from google.colab import userdata
from pinecone import Pinecone, ServerlessSpec

PINECONE_API_KEY = userdata.get("PINECONE_API_KEY")

pc = Pinecone(api_key=PINECONE_API_KEY)

index_name = "sci-judgments-hybrid"

if not pc.has_index(index_name):
    pc.create_index(
        name=index_name,
        dimension=4096,
        metric="dotproduct",
        spec=ServerlessSpec(
            cloud="aws",
            region="us-east-1"
        )
    )

print("Index created!")
print(pc.describe_index(index_name))
```

```python
from google.colab import userdata
import requests

NVIDIA_API_KEY = userdata.get("NVIDIA_API_KEY")

def generate_embedding(text):
    response = requests.post(
        "https://integrate.api.nvidia.com/v1/embeddings",
        headers={
            "Authorization": f"Bearer {NVIDIA_API_KEY}",
            "Content-Type": "application/json"
        },
        json={
            "model": "nvidia/nemotron-3.5-embed-8b",
            "input": [text],
            "input_type": "passage",
            "encoding_format": "float"
        }
    )

    response.raise_for_status()
    return response.json()["data"][0]["embedding"]
```

```python
text = "The accused is entitled to the benefit of doubt."

dense_vector = generate_embedding(text)

print("Dimensions:", len(dense_vector))
print("First 5 values:", dense_vector[:5])
```

```python
from google.colab import userdata
from pinecone import Pinecone
from pinecone_text.sparse import BM25Encoder
import requests
import uuid

# API keys
NVIDIA_API_KEY = userdata.get("NVIDIA_API_KEY")
PINECONE_API_KEY = userdata.get("PINECONE_API_KEY")

# Initialize Pinecone
pc = Pinecone(api_key=PINECONE_API_KEY)
index = pc.Index("sci-judgments-hybrid")

# Load your previously fitted BM25 encoder
bm25 = BM25Encoder()
bm25.load("bm25_params.json")


def upsert_judgment(text, metadata=None):

    # 1. Generate NVIDIA dense embedding
    response = requests.post(
        "https://integrate.api.nvidia.com/v1/embeddings",
        headers={
            "Authorization": f"Bearer {NVIDIA_API_KEY}",
            "Content-Type": "application/json"
        },
        json={
            "model": "nvidia/nemotron-3.5-embed-8b",
            "input": [text],
            "input_type": "passage",
            "encoding_format": "float"
        }
    )

    response.raise_for_status()
    dense_vector = response.json()["data"][0]["embedding"]

    # 2. Generate BM25 sparse embedding
    sparse_vector = bm25.encode_documents(text)

    # 3. Prepare metadata
    record_metadata = metadata.copy() if metadata else {}
    record_metadata["text"] = text

    # 4. Upsert both vectors
    index.upsert(
        vectors=[{
            "id": str(uuid.uuid4()),
            "values": dense_vector,
            "sparse_values": sparse_vector,
            "metadata": record_metadata
        }]
    )

    return "Successfully upserted judgment chunk!"
```

```python
result = upsert_judgment(
    text="The accused is entitled to the benefit of doubt...",
    metadata={
        "case_id": "2023_INSC_123",
        "case_title": "Example vs State",
        "paragraph_number": 12
    }
)

print(result)
```
